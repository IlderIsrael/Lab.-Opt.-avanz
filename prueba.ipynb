# Articulo Yoshitake2004. Viscosity Measure

```python
import math
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import erfc

# Parámetros físicos del láser y fluidos
spot_diameter = 1E-4  # Ancho del haz láser (m)

# Viscosidades (Pa·s)
viscosity_100cSt = 0.0965
viscosity_1000cSt = 0.970
viscosity_water = 1.0016E-3

# Tensiones superficiales (N/m)
surface_tension_100cSt = 20.9E-3
surface_tension_1000cSt = 21.2E-3
surface_tension_water = 7.28E-2

# Cálculo de tiempos característicos sin corrección
tau_100cSt = (math.sqrt(2) * viscosity_100cSt * spot_diameter) / surface_tension_100cSt
tau_1000cSt = (math.sqrt(2) * viscosity_1000cSt * spot_diameter) / surface_tension_1000cSt
tau_water = (math.sqrt(2) * viscosity_water * spot_diameter) / surface_tension_water

# Resultados impresos
print("For 100 cSt 1/Gamma = %.2f ms" % (round(tau_100cSt * 1000, 2)))
print("For 1000 cSt 1/Gamma = %.2f ms" % (round(tau_1000cSt * 1000, 2)))
print("For water 1/Gamma = %.4f ms" % (round(tau_water * 1000, 4)))

# Bloque para el análisis gráfico
density_fluid = 965  # kg/m³
kinematic_viscosity = np.linspace(1, 1E6, 100)  # en cSt

# Cálculo de tiempo sin corrección (en función de la viscosidad cinemática)
time_uncorrected = (math.sqrt(2) * kinematic_viscosity * 1E-6 * spot_diameter) / surface_tension_100cSt

# Cálculo del término de corrección
correction_term = ((density_fluid * (spot_diameter)**3) / (surface_tension_100cSt * 2 * math.pi))**(1/2)

# Tiempo corregido
time_corrected = time_uncorrected + correction_term

# Gráfica
plt.figure(figsize=(10, 6))
plt.plot(time_uncorrected * 1000, kinematic_viscosity, label='Sin corrección', color='blue')
plt.plot(time_corrected * 1000, kinematic_viscosity, label='Con corrección', color='red')
plt.title('Relación entre Tiempo y Viscosidad Cinemática')
plt.xlabel('Tiempo (ms)')
plt.ylabel('Viscosidad cinemática (cSt)')
plt.yscale("log")
plt.xscale("log")
plt.grid(True)
plt.legend()
plt.show()
```

# Para ele articulo citado, casner2001:

En el artículo \textit{``Giant Deformations of a Liquid-Liquid Interface Induced by the Optical Radiation Pressure''} de Casner y Delville, el número \textbf{Bo} se refiere al \textbf{número de Bond óptico}, una versión adaptada del clásico número de Bond que compara fuerzas gravitatorias con fuerzas de tensión superficial, pero aquí aplicado al contexto de la presión de radiación óptica.

\section*{Definición del número de Bond óptico (Bo)}

El número de Bond óptico se define como:

\[
\text{Bo} = \left( \frac{v_0}{l_C} \right)^2
\]

donde:

\begin{itemize}
  \item $v_0$ es el \textbf{radio del haz láser (beam waist)},
  \item $l_C$ es la \textbf{longitud capilar}, definida como:
  \[
  l_C = \sqrt{ \frac{\sigma}{(\rho_1 - \rho_2)g} }
  \]
  \item $\sigma$ es la \textbf{tensión superficial} entre las dos fases,
  \item $\rho_1$ y $\rho_2$ son las \textbf{densidades} de las dos fases líquidas,
  \item $g$ es la \textbf{aceleración gravitacional}.
\end{itemize}

\section*{Interpretación física}

El número de Bond óptico \textbf{Bo} indica la relación entre las fuerzas de flotación (gravedad) y las fuerzas de tensión superficial a la escala del haz láser. Dependiendo de su valor:

\begin{itemize}
  \item \textbf{Bo $\ll$ 1}: Dominan las fuerzas de tensión superficial, y la deformación del menisco es \textbf{no local} (la forma de la interfaz no sigue exactamente el perfil del haz).
  \item \textbf{Bo $\gg$ 1}: Dominan las fuerzas de flotación (gravedad), y la deformación se vuelve \textbf{local}, es decir, el perfil del menisco sigue el perfil del haz láser.
\end{itemize}

\section*{Uso en el artículo}

Los autores utilizan \textbf{Bo} como parámetro adimensional clave para unificar sus datos experimentales en una \textbf{curva maestra universal}. Al representar los resultados como función de \textbf{Bo}, pueden describir el comportamiento de la deformación del menisco inducida por presión de radiación láser en distintos regímenes de tensión superficial y tamaños de haz, desde superficies muy suaves hasta más rígidas.
